```python
from sqlalchemy import create_engine
```

```python
engine = create_engine("mysql+pymysql://root:Root@localhost:3306/sample")
```

```python
df = pd.read_csv("cleaned_ola_rides.csv")
```

```python
df.to_sql(
    name='booking_data',
    con=engine,
    if_exists='replace',   
    index=False,         
    chunksize=1000        
)

print("Table successfully created and data loaded!")
```

```python
df= pd.read_sql("SELECT * FROM booking_data", con=engine)
```

```python
df= pd.read_sql("""
SELECT 
    * 
FROM 
    booking_data 
WHERE 
    Booking_Status = 'Success';
""", con=engine)   
```

```python
df
```

```python
df= pd.read_sql("""
SELECT 
    Vehicle_Type,
    ROUND(AVG(Ride_Distance), 2) AS avg_ride_distance
FROM 
    booking_data
WHERE 
    Vehicle_Type IS NOT NULL
GROUP BY 
    Vehicle_Type
ORDER BY 
    avg_ride_distance DESC;
""", con=engine)
```

```python
df
```

```python
df= pd.read_sql("""
SELECT 
    COUNT(*) AS total_cancelled_by_customers
FROM 
    booking_data
WHERE 
    Booking_Status = 'Cancelled by Customer';
""", con=engine)    
```

```python
df
```

```python
df= pd.read_sql("""SELECT 
    Customer_ID,
    COUNT(*) AS total_rides
FROM 
    booking_data
WHERE 
    Customer_ID IS NOT NULL
GROUP BY 
    Customer_ID
ORDER BY 
    total_rides DESC
LIMIT 5;
""", con=engine)    
```

```python
df
```

```python
df= pd.read_sql("""SELECT 
    COUNT(*) AS total_driver_cancelled_rides
FROM 
    booking_data
WHERE 
    Incomplete_Rides_Reason LIKE '%%Personal%%' 
    AND Incomplete_Rides_Reason LIKE '%%Car%%';
""", con=engine)
```
